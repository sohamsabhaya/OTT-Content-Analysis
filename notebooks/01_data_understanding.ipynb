```python
import pandas as pd
import numpy as np
import os

# Define raw data file mapping
data_files = {
    'Netflix': '../data/raw/netflix.csv',
    'Amazon Prime': '../data/raw/amazon_prime.csv',
    'Disney+': '../data/raw/disney_plus.csv',
    'Hulu': '../data/raw/hulu.csv'
}

for name, path in data_files.items():
    print(f'{name}: exists = {os.path.exists(path)}')
```

## 1. Dimensions and Catalog Overview

```python
dfs = {}
overview_records = []

for name, path in data_files.items():
    df = pd.read_csv(path)
    dfs[name] = df
    overview_records.append({
        'Platform': name,
        'Total Rows': df.shape[0],
        'Total Columns': df.shape[1],
        'Duplicate Show IDs': df['show_id'].duplicated().sum(),
        'Duplicate Titles': df['title'].duplicated().sum()
    })

overview_df = pd.DataFrame(overview_records)
overview_df
```

## 2. Schema Comparison Across Platforms

```python
schema_comp = pd.DataFrame({
    platform: pd.Series(df.columns)
    for platform, df in dfs.items()
})
schema_comp
```

## 3. Sample Records and Data Types

```python
for platform, df in dfs.items():
    print('='*50)
    print(f'Platform: {platform}')
    print('='*50)
    print(df.dtypes)
    print('\nFirst 2 Records:')
    print(df[['show_id', 'type', 'title', 'release_year', 'rating', 'duration']].head(2))
    print()
```

## 4. Missing Value Assessment (Data Quality Audit)

```python
null_reports = []
for platform, df in dfs.items():
    n_counts = df.isnull().sum()
    n_pct = (n_counts / len(df) * 100).round(2)
    report = pd.DataFrame({'Missing_Count': n_counts, 'Missing_Pct': n_pct})
    report['Platform'] = platform
    null_reports.append(report[report['Missing_Count'] > 0])

pd.concat(null_reports)
```

## 5. Summary Findings
- All four catalogs follow the identical 12-column schema.
- Missing values are heavily concentrated in `director`, `cast`, and `country`.
- Netflix and Hulu exhibit duration values mistakenly captured in `rating`.
- The findings inform the cleaning strategy in Notebook 02.
