## Key Findings Summary
1. **Movies Dominate:** 71.4% of total OTT content is Movies, though Hulu is TV show heavy (53.7%).
2. **Catalog Scale:** Amazon Prime has the largest catalog (9,668), followed by Netflix (8,807).
3. **Top Genres:** Drama and Comedy account for the largest share across all platforms.
4. **Geographic Concentration:** The United States, India, and the United Kingdom produce the vast majority of titles.
5. **Movie Duration:** Bell-shaped distribution centered around a median of 94 minutes.
6. **TV Show Retention:** 64.2% of shows conclude or are cancelled after 1 season; 35.8% achieve renewal.
